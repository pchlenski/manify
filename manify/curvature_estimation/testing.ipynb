```python
tree_distances = torch.tensor([
    [0, 1, 2, 3, 3],
    [1, 0, 1, 2, 2],
    [2, 1, 0, 1, 1],
    [3, 2, 1, 0, 2],
    [3, 2, 1, 2, 0]
], dtype=torch.float32)

# Create a distance matrix for a grid (has positive hyperbolicity)
grid_size = 3
n_grid = grid_size * grid_size
grid_distances = torch.zeros((n_grid, n_grid), dtype=torch.float32)

# Fill grid distances with Manhattan distances
for i in range(n_grid):
    row_i, col_i = i // grid_size, i % grid_size
    for j in range(n_grid):
        row_j, col_j = j // grid_size, j % grid_size
        grid_distances[i, j] = abs(row_i - row_j) + abs(col_i - col_j)

# Calculate delta-hyperbolicity for tree
delta_tree, delta_rel_tree = delta_hyperbolicity_iterative(tree_distances)
print(f"Tree (Iterative): δ = {delta_tree:.6f}, relative δ = {delta_rel_tree:.6f}")

print(f'approx: {approx_delta_hyperbolicity(tree_distances)}')

delta_tree, delta_rel_tree = delta_hyperbolicity_parallel(tree_distances)
print(f"Tree (Parallel): δ = {delta_tree:.6f}, relative δ = {delta_rel_tree:.6f}")

# Calculate delta-hyperbolicity for grid
delta_grid, delta_rel_grid = delta_hyperbolicity_iterative(grid_distances)
print(f"Grid (Iterative): δ = {delta_grid:.6f}, relative δ = {delta_rel_grid:.6f}")

delta_grid, delta_rel_grid = delta_hyperbolicity_parallel(grid_distances)
print(f"Grid (Parallel): δ = {delta_grid:.6f}, relative δ = {delta_rel_grid:.6f}")

print(f'approx: {approx_delta_hyperbolicity(grid_distances)}')
```
